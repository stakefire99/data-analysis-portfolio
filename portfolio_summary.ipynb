# 📊 Data Analysis Portfolio – Master Summary
## All 5 Projects Overview

```python
import sys, os
sys.path.insert(0, os.path.abspath('.'))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings; warnings.filterwarnings('ignore')
print("✅ Imports OK")
```

```python
projects = [
    {'name':'Retail Sales',   'folder':'project1_sales_analysis',       'csv':'processed_sales_data.csv',      'value_col':'SalesAmount'},
    {'name':'Healthcare',     'folder':'project2_healthcare_analysis',   'csv':'processed_healthcare_data.csv', 'value_col':'TreatmentCost'},
    {'name':'Sports',         'folder':'project3_sports_analytics',      'csv':'processed_sports_data.csv',     'value_col':'Points'},
    {'name':'Finance',        'folder':'project4_financial_analysis',    'csv':'stock_returns.csv',             'value_col':None},
    {'name':'E-commerce',     'folder':'project5_ecommerce_analytics',   'csv':'transaction_data.csv',          'value_col':'TotalAmount'},
]
summary = []
for p in projects:
    path = os.path.join(p['folder'], p['csv'])
    if os.path.exists(path):
        df = pd.read_csv(path)
        val = df[p['value_col']].sum() if p['value_col'] and p['value_col'] in df.columns else 0
        summary.append({'Project':p['name'],'Records':len(df),'Columns':len(df.columns),'Value':val})
    else:
        summary.append({'Project':p['name'],'Records':0,'Columns':0,'Value':0})

summary_df = pd.DataFrame(summary)
print(summary_df.to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(14,6))
colors = ['#2E86AB','#A23B72','#F18F01','#C73E1D','#6A4C93']

axes[0].bar(summary_df['Project'], summary_df['Records'], color=colors)
axes[0].set_title('Records Analysed per Project', fontweight='bold')
axes[0].set_ylabel('Records'); axes[0].tick_params(axis='x', rotation=20)

axes[1].pie(summary_df['Records']+1, labels=summary_df['Project'],
            autopct='%1.1f%%', colors=colors)
axes[1].set_title('Data Distribution', fontweight='bold')
plt.tight_layout()
plt.savefig('master_dashboard.png', dpi=150, bbox_inches='tight')
plt.show()
print("✅ Saved: master_dashboard.png")
```
